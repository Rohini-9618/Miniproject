```python
# 1. Before cleaning missing values check 
df.isnull().sum()
```

```python
df.dropna(inplace=True)

# 2. Missing values filling & data cleaning
print("\nMissing values after cleaning:")
missing_after = df.isnull().sum()
print(missing_after[missing_after > 0])

# 3. Cleaning-nu seshamulla missing values check 
print("\nCleaned dataset shape (Rows, Columns):", df.shape)
```

```python
initial_rows = df.shape[0]
df.drop_duplicates(inplace=True)
print(f"Removed duplicates: {initial_rows - df.shape[0]} rows")
```

```python
filtered_df = df[df['DEPT_NM'] == 'Austin Public Library']
aggregated_df =filtered_df.groupby('DEPT_NM')['AMOUNT'].sum()
print(aggregated_df)
```

```python
df['AMOUNT'] = pd.to_numeric(df['AMOUNT'].astype(str).str.replace(',', ''), errors='coerce')
df['CHK_EFT_ISS_DT'] = pd.to_datetime(df['CHK_EFT_ISS_DT'], errors='coerce')
df['TRANS_YEAR'] = df['CHK_EFT_ISS_DT'].dt.year
df['TRANS_MONTH'] = df['CHK_EFT_ISS_DT'].dt.month
df['DAY_OF_WEEK'] = df['CHK_EFT_ISS_DT'].dt.day_name()
def get_amount_category(amt):
    if pd.isna(amt):
        return 'Unknown'
    elif amt < 1000:
        return 'Small'
    elif amt <= 50000:
        return 'Medium'
    else:
        return 'Large'

df['AMOUNT_CATEGORY'] = df['AMOUNT'].apply(get_amount_category)
print(df[['AMOUNT', 'TRANS_YEAR', 'TRANS_MONTH', 'DAY_OF_WEEK', 'AMOUNT_CATEGORY']].head())
```

```python
# Outlier Handling using IQR method (Removing/Filtering negative and extreme values)
Q1 = df['AMOUNT'].quantile(0.25)
Q3 = df['AMOUNT'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = max(0, Q1 - 1.5 * IQR)
upper_bound = Q3 + 1.5 * IQR

# Filter the dataframe to keep only valid amounts between lower_bound and upper_bound
initial_rows = len(df)
df = df[(df['AMOUNT'] >= lower_bound) & (df['AMOUNT'] <= upper_bound)]

print(f"Removed outliers. Rows before: {initial_rows}, Rows after: {len(df)}")
print(f"New Lower bound: {lower_bound}, Upper bound: {upper_bound}")
```
